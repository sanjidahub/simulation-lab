```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Scalar and custom array
scalar = 3.7
arr = np.array([1.2, 2.5, 3.7, 4.8, 5.5, 6.9])

print("Scalar:", scalar)
print("Array:", arr)

print("\n--- Trigonometric Functions ---")
print("sin:", np.sin(arr))
print("cos:", np.cos(arr))
print("tan:", np.tan(arr))

# Clip values to avoid arcsin/arccos domain errors
clipped_arr = np.clip(arr / 7, -1, 1)

print("\n--- Inverse Trigonometric Functions ---")
print("arcsin:", np.arcsin(clipped_arr))
print("arccos:", np.arccos(clipped_arr))
print("arctan:", np.arctan(arr))

print("\n--- Exponential & Logarithmic Functions ---")
print("exp:", np.exp(arr))
print("ln:", np.log(arr))
print("log10:", np.log10(arr))
print("log2:", np.log2(arr))

print("\n--- Absolute Value & Square Root ---")
print("Absolute:", np.abs(arr))
print("Square root:", np.sqrt(arr))

print("\n--- Remainder when divided by 2 ---")
print(np.remainder(arr, 2))

print("\n--- Rounding ---")
print("Round:", np.round(arr))
print("Floor:", np.floor(arr))
print("Ceil:", np.ceil(arr))
```

```python
# Create a custom 3×2 matrix
matrix_3x2 = np.array([
    [0.5, 1.2],
    [2.5, 3.7],
    [4.2, 5.5]
])

print("3×2 Matrix:")
print(matrix_3x2)

print("\n--- Trigonometric Functions ---")
print("sin:")
print(np.sin(matrix_3x2))

print("\ncos:")
print(np.cos(matrix_3x2))

print("\ntan:")
print(np.tan(matrix_3x2))

# Clip values for arcsin and arccos
clipped_matrix = np.clip(matrix_3x2 / 6, -1, 1)

print("\narcsin:")
print(np.arcsin(clipped_matrix))

print("\narccos:")
print(np.arccos(clipped_matrix))

print("\narctan:")
print(np.arctan(matrix_3x2))
```

```python
print("--- Exponential Functions ---")
print("Exponential:")
print(np.exp(matrix_3x2))

print("\n--- Logarithmic Functions ---")
print("Natural log:")
print(np.log(matrix_3x2))

print("\nLog base 10:")
print(np.log10(matrix_3x2))

print("\nLog base 2:")
print(np.log2(matrix_3x2))

print("\n--- Absolute Value & Square Root ---")
print("Absolute value:")
print(np.abs(matrix_3x2))

print("\nSquare root:")
print(np.sqrt(matrix_3x2))

print("\n--- Remainder when divided by 2 ---")
print(np.remainder(matrix_3x2, 2))

print("\n--- Rounding ---")
print("Round:")
print(np.round(matrix_3x2))

print("\nFloor:")
print(np.floor(matrix_3x2))

print("\nCeil:")
print(np.ceil(matrix_3x2))
```

```python
# Create a custom 2×3 matrix
matrix_2x3 = np.array([
    [2.1, 4.7, 1.5],
    [6.8, 3.2, 5.9]
])

print("2×3 Matrix:")
print(matrix_2x3)

# Maximum and minimum
max_value = np.max(matrix_2x3)
min_value = np.min(matrix_2x3)

max_index = np.unravel_index(
    np.argmax(matrix_2x3), matrix_2x3.shape
)

min_index = np.unravel_index(
    np.argmin(matrix_2x3), matrix_2x3.shape
)

print("\nMaximum value:", max_value)
print("Maximum index:", max_index)

print("\nMinimum value:", min_value)
print("Minimum index:", min_index)

# Vector length
print("\nVector length:")
print(np.linalg.norm(matrix_2x3))

# Row-wise descending sort
row_sorted = np.sort(matrix_2x3, axis=1)[:, ::-1]

print("\nRow-wise descending sort:")
print(row_sorted)

# Entire matrix descending sort
entire_sorted = np.sort(matrix_2x3.flatten())[::-1]

print("\nEntire matrix descending sort:")
print(entire_sorted)

# Sum and product
print("\nSum:", np.sum(matrix_2x3))
print("Product:", np.prod(matrix_2x3))

# Statistics
print("\nMedian:", np.median(matrix_2x3))
print("Mean:", np.mean(matrix_2x3))
print("Standard Deviation:", np.std(matrix_2x3))
```

```python
# Make the graph size
plt.figure(figsize=(7, 4))

# Show the matrix as a heatmap
sns.heatmap(
    matrix_2x3,      # Our 2x3 matrix
    annot=True,      # Show numbers in the boxes
    fmt=".2f",       # Show numbers up to 2 decimal places
    cmap="viridis"   # Choose a color style
)

# Give the graph a title
plt.title("2x3 Matrix Heatmap")

# Give names to the axes
plt.xlabel("Columns")
plt.ylabel("Rows")

# Show the graph
plt.show()


# Create a random 2x3 matrix
random_matrix = np.random.uniform(1, 10, size=(2, 3))

# Print the random matrix
print("Random 2x3 Matrix:")
print(random_matrix)
```
