```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

Uniform Random Numbers

```python

numbers = np.random.rand(10)

print("1×10 Random Vector:")
print(numbers)

# Histogram
plt.hist(numbers, bins=5, edgecolor='black')
plt.title("Uniform Random Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

Normally Distributed Random Numbers:

```python
numbers = np.random.normal(0, 1, 500)

print("First 10 numbers:")
print(numbers[:10])

# Histogram
plt.hist(numbers, bins=20, edgecolor='black')
plt.title("Standard Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

Random Integers WITH Fixed Seed

```python

# WITH fixed seed
np.random.seed(10)
matrix1 = np.random.randint(1, 51, (3, 3))

# WITHOUT seed
matrix2 = np.random.randint(1, 51, (3, 3))

print("With Fixed Seed (10):")
print(matrix1)

print("\nWithout Fixed Seed:")
print(matrix2)

print("\nObservation:")
print("Fixed seed gives the same values every time.")
print("Without seed, the values change each time the code runs.")
```

One Sample T-Test

```python
import numpy as np
from scipy.stats import ttest_1samp
import matplotlib.pyplot as plt

# Given data and hypothesized mean
data = [5.2, 5.4, 5.1, 5.6, 5.3, 5.8]
mean = 5.5

# One-sample t-test
t, p = ttest_1samp(data, mean)

print("T-statistic:", round(t, 4))
print("P-value:", round(p, 4))

# Interpretation
if p < 0.05:
    print("Reject H0: Significant difference.")
else:
    print("Fail to reject H0: No significant difference.")

# Visualization
plt.hist(data, bins=5, edgecolor='black')
plt.axvline(mean, linestyle='--', label='Hypothesized Mean = 5.5')
plt.title("One Sample T-Test")
plt.xlabel("Data")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
