```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sclar= 4.3
arr = np.array([1.2, -2.5, 3.8, -4.1, 5.6, -6.7, 7.9, 8.4],dtype=float)
sqrt_values = np.sqrt(arr)
abs_values=np.abs(arr)
exp_values=np.exp(arr)
log_values=np.log(arr)
sin_values = np.sin(arr)
cos_values = np.cos(arr)
tan_values = np.tan(arr)
clipped=np.clip(arr/10,-1,1)
asin_values=np.arcsin(np.clip(arr/np.max(clipped),-1,1))
acos_values=np.arccos(np.clip(arr/np.max(clipped),-1,1))
atan_values=np.arctan(clipped)



print("add: \n",arr+sclar)
print("sub: \n",arr-sclar)
print("mul: \n",arr*sclar)
print("div: \n",arr/sclar)
print("sqrt: ",sqrt_values)
print("abs: ",abs_values)
print("exp: ",exp_values)
print("log: ",log_values)
print("sin: ",sin_values)
print("cos: ",cos_values)
print("tan: ",tan_values)
print("asin: ",asin_values)
print("acos: ",acos_values)
print("atan: ",atan_values)
```

```python
matrix = np.array([[1.5,-2.3,3.7],[-4.2,5.1,-6.8],[7.4,-8.9,9.2],[-1.7,2.6,-3.5]])
sin_values=np.sin(matrix)
cos_values = np.cos(matrix)
tan_values = np.tan(matrix)
sin_values=np.arcsin(np.clip(matrix/np.max(matrix),-1,1))
acos_values=np.arccos(np.clip(matrix/np.max(matrix),-1,1))
atan_values=np.arctan(matrix)
exp_values=np.exp(matrix)
log_values=np.log(matrix)
log_values10=np.log10(matrix)
abs_values=np.abs(matrix)
sqrt_values=np.sqrt(matrix)
sqaure=np.square(matrix)
cube_root=np.cbrt(matrix)
rem_values=np.remainder(matrix,3)
round_values=np.round(matrix)
floor_values=np.floor(matrix)
ceil_values=np.ceil(matrix)

print("sin:\n ",sin_values)
print("cos:\n ",cos_values)
print("tan: \n",tan_values)
print("asin: \n",asin_values)
print("acos: \n",acos_values)
print("atan:\n ",atan_values)
print("exp: \n",exp_values)
print("log: \n",log_values)
print("log: \n",log_values10)
print("abs: \n",abs_values)
print("sqrt: \n",sqrt_values)
print("Sqaure: ",sqaure)
print("CubeRoot: ",cube_root)
print("rem:\n ",rem_values)
print("round:\n ",round_values)
print("floor:\n ",floor_values)
print("ceil: \n",ceil_values)
```
